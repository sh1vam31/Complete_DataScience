## 1 Inspecting Data
Quick ways to understand your dataset:

- head(n) → first n rows (default 5)
- tail(n) → last n rows
- info() → columns, non-null counts, and dtypes
- shape → (rows, columns)
- dtypes → data type of each column

```python
import pandas as pd

a = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [25, 30, 35, 40, 22],
    "Score": [88, 92, 79, 85, 95]
}

b = pd.DataFrame(a)
b
print()
print(b.head())
print()
print(b.head(3))
print()
print(b.tail())
print()
print(b.tail(2))
print()
print(b.shape)
print()
print(b.info())
print()
print(b.dtypes)
```

## 2 Selecting Columns & Rows
Ways to access parts of a DataFrame:

- Single column → `df['col']`
- Multiple columns → `df[['col1','col2']]`
- Label-based selection → `df.loc[row_label, col_label]`
- Integer-index selection → `df.iloc[row_idx, col_idx]` Tip: : means "all rows/columns" in that dimension.

```python
a = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [25, 30, 35, 40, 22],
    "Score": [88, 92, 79, 85, 95]
}

b = pd.DataFrame(a)
print(b)
print()
print(b["Name"])
print()
print(b[["Name", "Age"]])     # whenever we use the fancy indexing we have to use the 2 square brackats
print()
print(b.loc[1])
print()
print(b.loc[[2, 3]])
print()
print(b.loc[4,"Score"])
print()
print(b.loc[1:4, ["Name", "Age"]])
print()
print(b.iloc[2, 2])
```

## 3 Index & Columns Overview
- `df.index` → row labels
- `df.columns` → column names
- `df.values` → underlying NumPy array

```python
print(b)
print()
print(b.index)
print()
print(b.columns)
print()
print(b.values)
```

## 4 Renaming Columns & Setting an Index
- Rename columns → `df.rename(columns={'old':'new'}, inplace=False)`
- Set a column as `index → df.set_index('column_name', inplace=False)`
- Reset index → `df.reset_index(inplace=False)` If you don't use `inplace=True`, remember to assign the result back to a variable.

```python
print(b)
print()
c = b.rename(columns = {"Score": "marks"}, inplace = False)
print(c)
print("\nThe Original dataframe")
print(b)
print()
c = b.set_index("Name")
print(c)
print()
b.reset_index(inplace = False)
print(b)
```

## Handling Missing Values
Real-world datasets often have missing data. Pandas offers:

- isna() / isnull() → detect missing values (returns True/False)
- dropna() → remove missing values (rows/columns)
- fillna() → replace missing values with a given value or method (e.g., forward fill)
- Always inspect how much data is missing before deciding to drop or fill.

```python
import numpy as np
a = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [25, None, 35, 40, None],
    "City": ["Delhi", "Mumbai", None, "Bangalore", "Chennai"],
    "Score": [85, 91, np.nan, 88, 95]
}

df = pd.DataFrame(a)
df.isna()
```

```python
df.isna().sum()
```

```python
df.dropna()
```

```python
print(df)
d = df.copy()
print("\n After filling all the values")
d["Age"] = d["Age"].fillna(d["Age"].mean())
d["City"] = d["City"].fillna(d["City"].mode())
d["Score"] = d["Score"].fillna(d["Score"].mean())
print(d)

d.isna().sum()
```

## Changing Data Types (astype)
- Convert a column to a different type using `astype()`.
- Examples: converting strings to numbers, integers to floats, or columns to category.
- Be careful: make sure the data is compatible with the new type!

```python
a = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [25, 30, 35, 40, 22],
    "Score": [88, 92, 79, 85, 95]
}

df = pd.DataFrame(a)
print(df)
print("Change the Score to the float dataType")
df["Score"] = df["Score"].astype(float)
print(df)
print(df.dtypes)
```

- Useful for cleaning labels, fixing typos, or mapping categories.

```python
c["Score"] = c["Score"].replace({88.0:90.0, 79.0:66.0})
c
```

## Dropping Rows & Columns `(drop)`
- `drop(labels, axis=0)` → drop rows by index labels. axis = 0 means it will search the label in the rows
- `drop(labels, axis=1)` → drop columns by name. axis = 0 means it will search the label in the columns
- Use `inplace=True` if you want to modify the DataFrame directly.

```python
c.drop("Name", axis=1)
c.drop([0,2], axis=0)
```

## Detecting & Removing Duplicates
- `duplicated()` → returns a Boolean Series marking duplicate rows.
- `drop_duplicates()` → remove duplicate rows.
- You can specify `subset (columns)` and keep` ('first', 'last', or False)`.

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "Alice", "Bob", "David"],
    "Age": [25, 30, 35, 25, 30, 40],
    "Score": [88, 92, 79, 88, 92, 85]
}

df = pd.DataFrame(data)
df
```

```python
print(df.duplicated())
print()
print(df.duplicated().sum())
```

```python
df.drop_duplicates()
```

```python
# 3. Remove duplicates (keep first occurrence)
df_first = df.drop_duplicates()
print("\nAfter dropping duplicates (keep='first'):")
print(df_first)

# 4. Remove duplicates (keep last occurrence)
df_last = df.drop_duplicates(keep="last")
print("\nAfter dropping duplicates (keep='last'):")
print(df_last)

# 5. Remove all duplicates (keep=False)
df_none = df.drop_duplicates(keep=False)
print("\nAfter removing all duplicates (keep=False):")
print(df_none)

# 6. Remove duplicates based on a subset of columns (e.g., Name only)
df_subset = df.drop_duplicates(subset=["Name"])
print("\nAfter dropping duplicates based on 'Name' only:")
print(df_subset)
```
