## Handling Missing Values
Real-world datasets often have missing data. Pandas offers:

- isna() / isnull() → detect missing values (returns True/False)
- dropna() → remove missing values (rows/columns)
- fillna() → replace missing values with a given value or method (e.g., forward fill)
- Always inspect how much data is missing before deciding to drop or fill.

```python
import numpy as np
a = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [25, None, 35, 40, None],
    "City": ["Delhi", "Mumbai", None, "Bangalore", "Chennai"],
    "Score": [85, 91, np.nan, 88, 95]
}

df = pd.DataFrame(a)
df.isna()
```

```python
df.isna().sum()
```

```python
df.dropna()
```

```python
print(df)
d = df.copy()
print("\n After filling all the values")
d["Age"] = d["Age"].fillna(d["Age"].mean())
d["City"] = d["City"].fillna(d["City"].mode())
d["Score"] = d["Score"].fillna(d["Score"].mean())
print(d)

d.isna().sum()
```

## Detecting & Removing Duplicates
- `duplicated()` → returns a Boolean Series marking duplicate rows.
- `drop_duplicates()` → remove duplicate rows.
- You can specify `subset (columns)` and keep` ('first', 'last', or False)`.

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "Alice", "Bob", "David"],
    "Age": [25, 30, 35, 25, 30, 40],
    "Score": [88, 92, 79, 88, 92, 85]
}

df = pd.DataFrame(data)
df
```

```python
print(df.duplicated())
print()
print(df.duplicated().sum())
```

```python
df.drop_duplicates()
```

```python
# 3. Remove duplicates (keep first occurrence)
df_first = df.drop_duplicates()
print("\nAfter dropping duplicates (keep='first'):")
print(df_first)

# 4. Remove duplicates (keep last occurrence)
df_last = df.drop_duplicates(keep="last")
print("\nAfter dropping duplicates (keep='last'):")
print(df_last)

# 5. Remove all duplicates (keep=False)
df_none = df.drop_duplicates(keep=False)
print("\nAfter removing all duplicates (keep=False):")
print(df_none)

# 6. Remove duplicates based on a subset of columns (e.g., Name only)
df_subset = df.drop_duplicates(subset=["Name"])
print("\nAfter dropping duplicates based on 'Name' only:")
print(df_subset)
```

## Concept of Split–Apply–Combine
- Split the data into groups based on some criteria.
- Apply a function (e.g., sum, mean) to each group.
- Combine the results into a new DataFrame or Series.
- This is the core idea behind `groupby`.

```python
import pandas as pd

data = {
    "Employee": ["John", "Sarah", "Mike", "Anna", "Tom", "Laura", "Steve", "Kate"],
    "Department": ["HR", "IT", "IT", "Finance", "Finance", "HR", "IT", "Finance"],
    "Salary": [50000, 70000, 72000, 65000, 60000, 52000, 75000, 58000]
}

df = pd.DataFrame(data)
print("\nAverage salary by Department:")
print(df.groupby("Department")["Salary"].mean())
print("\nNumber of employees by Department:")
print(df.groupby("Department")["Employee"].count())
print("\nSalary stats (mean, max, min) by Department:")
print(df.groupby("Department")["Salary"].agg(["mean", "max", "min"]))
```

## Multiple Aggregations on Different Columns
- With `agg()`, you can specify different functions for each column.
- Example: `df.groupby('Dept').agg({'Salary':'mean', 'Age':'max'})`.

```python
data = {
    "Employee": ["John", "Sarah", "Mike", "Anna", "Tom", "Laura", "Steve", "Kate"],
    "Department": ["HR", "IT", "IT", "Finance", "Finance", "HR", "IT", "Finance"],
    "Salary": [50000, 70000, 72000, 65000, 60000, 52000, 75000, 58000],
    "Age": [28, 32, 30, 27, 35, 29, 31, 26]
}

df = pd.DataFrame(data)
df.groupby("Department").agg({
    "Salary": ["max", "min"],
    "Age" : ["mean", "max"]
})
```

## Pivot Tables & Crosstab
- `pivot_table()` summarizes data like Excel pivot tables.
- `crosstab()` shows frequency counts of combinations of factors.
- Both are powerful for summarizing and comparing groups

```python
df
print("\nPivot table: Average Salary by Department")
print(df.pivot_table(values = "Salary", index = "Department", aggfunc = "mean"))
print("\nPivot table: Average Salary & Age by Department")
print(df.pivot_table(values = ["Salary" , "Age"], index = "Department", aggfunc = "mean"))

# Crosstab: number of employees by Department and Age group
# Let's create an AgeGroup column
df["AgeGroup"] = pd.cut(df["Age"], bins=[25, 30, 35, 40], labels=["25-30", "31-35", "36-40"])
crosstab_result = pd.crosstab(df["Department"], df["AgeGroup"])
print("\nCrosstab: Employees by Department and Age Group")
print(crosstab_result)
```

## Using `concat()` for Stacking Data
- `pd.concat()` combines DataFrames vertically (rows) or horizontally (columns).
- Use `axis=0` for vertical stacking (default), axis=1 for horizontal.
- `ignore_index=True` can reset the index after concatenation.

```python
import pandas as pd
# Dataset 1: Product details
df1 = pd.DataFrame({
    "ProductID": [101, 102, 103],
    "ProductName": ["Pen", "Notebook", "Marker"],
    "Price": [10, 50, 30]
})

# Dataset 2: More products (same columns as df1)
df2 = pd.DataFrame({
    "ProductID": [104, 105],
    "ProductName": ["Pencil", "Eraser"],
    "Price": [5, 8]
})

# Dataset 3: Additional info (different columns)
df3 = pd.DataFrame({
    "Stock": [100, 60, 80],
    "Rating": [4.5, 4.7, 4.8]
})

print("Original DataFrames:")
print("\nDF1:\n", df1)
print("\nDF2:\n", df2)
print("\nDF3:\n", df3)

pd.concat([df1, df2], axis = 0, ignore_index=True)
```

```python
pd.concat([df1, df3], axis = 1)
```

##  Using `merge()` for SQL-Style Joins
- `merge()` is similar to SQL joins.
- Specify `on` or `left_on`/`right_on` to choose key columns.
- Join types: `inner` (default), `left`, `right`, `outer`.

```python
products = pd.concat([df1, df2], axis=0, ignore_index=True)
print("\n product data")
print(products)

sales = pd.DataFrame({
    "ProductID": [101, 102, 104, 106],  # 106 doesn't exist in products
    "UnitsSold": [500, 200, 150, 50]
})
print("\n sales data")
print(sales)

print("\nInner Join (only matches):")
print(pd.merge(products, sales, on = "ProductID", how = "inner"))    # inner -> only the common values of product ID will be shown in the table
print("\nOuter Join (all data):")
print(pd.merge(products, sales, on = "ProductID", how = "outer") )   # outer -> all teh values of product ID will be shown in the table
print("\nLeft Join (all products):")
print(pd.merge(products, sales, on = "ProductID", how = "left") )
print("\nRight Join (all sales):")
print(pd.merge(products, sales, on = "ProductID", how = "right") )
```

## Using `join()` with Index Alignment
- `join()` joins DataFrames on their index or on a key column.
- By default, joins on index; can specify `on='col'`.
- Useful for adding columns to an existing DataFrame.

```python
products = pd.concat([df1, df2], ignore_index=True)
print("\n product data")
print(products)

stock = pd.DataFrame({
    "Stock": [100, 60, 80, 120, 150],
}, index=[101, 102, 103, 104, 105])      # ProductID as index
print("\n stock data")
print(stock)

p_data = products.set_index("ProductID")
print(p_data)
print("\n Joined data")
p_data.join(stock, on = "ProductID", how="outer")
```

```python
# Sales data
sales = pd.DataFrame({
    "ProductID": [201, 202, 203, 204],
    "UnitsSold": [500, 300, 400, 250]
})

# Revenue data (missing ProductID 204, extra 205)
revenue = pd.DataFrame({
    "ProductID": [201, 202, 203, 205],
    "Revenue": [7500, 4500, 6800, 1200]
})

print("Sales Data:")
print(sales)
print("\nRevenue Data:")
print(revenue)

# Merge sales and revenue to see performance
merged_perf = pd.merge(sales, revenue, on="ProductID", how="outer")
print("\nMerged Sales & Revenue (outer join):")
print(merged_perf)
```

## Applying Custom Functions with `apply()` / `map()`
- `apply()` lets you apply a custom function across rows or columns.
- `map()` is for applying functions element-wise to a Series.
- Both are handy for transformations, calculations, and cleaning data.

```python
import pandas as pd
import numpy as np
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Math": [88, 75, 92, 60, 85],
    "Science": [90, 80, 78, 70, 95]
}

df = pd.DataFrame(data)
df["total"] = df[["Math", "Science"]].apply(lambda r: r.sum(), axis = 1)

col_mean = df[["Math", "Science"]].apply(lambda c: c.mean(), axis = 0)
print(col_mean)

def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

df["Grade"] = df["Math"].apply(get_grade)
print(df)

df["Name"] = df["Name"].map(str.lower)
print(df)

df["Math"] = df["Math"].map(lambda x: x + 5)
print(df)
```
